```python
import pandas as pd

file_path = r"C:\Users\siddh\OneDrive\Documents\student-mat.csv.xlsx"

# Read the Excel file
raw_df = pd.read_excel(file_path, header=None)

# Split the first column wherever a semicolon appears
df = raw_df.iloc[:, 0].str.split(";", expand=True)

# Set the first row as column names
df.columns = df.iloc[0]
df = df.iloc[1:].reset_index(drop=True)

print("Dataset shape:", df.shape)
print("Column names:")
print(df.columns.tolist())

display(df.head())
```

```python
# Remove extra double quotes and spaces from text values
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip().str.strip('"')

# Check the cleaned data
display(df.head())

print("Dataset shape:", df.shape)
print("Gender values:", df["sex"].unique())
print("School values:", df["school"].unique())
```

```python
print("Missing values in each column:")
print(df.isnull().sum().sort_values(ascending=False).head(10))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Remove the unwanted column named '0', if it exists
if "0" in df.columns:
    df = df.drop(columns=["0"])

print("Updated dataset shape:", df.shape)
print("Column names:")
print(df.columns.tolist())
```

```python
# Convert numerical columns from text to numbers

numeric_cols = [
    "age", "Medu", "Fedu", "traveltime", "studytime",
    "failures", "famrel", "freetime", "goout",
    "Dalc", "Walc", "health", "absences", "G1", "G2", "G3"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Updated data types:")
print(df[numeric_cols].dtypes)

print("\nNumerical summary:")
display(df[numeric_cols].describe())
```

```python
# 1. Final grade distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="G3", bins=20, kde=True)
plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()

# 2. Attendance: absences vs final grade
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="absences", y="G3")
plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()

# 3. Parental education
parent_education = df.groupby(["Medu", "Fedu"])["G3"].mean().unstack()
display(parent_education)

plt.figure(figsize=(8, 5))
sns.heatmap(parent_education, annot=True, fmt=".1f", cmap="YlGnBu")
plt.title("Average Final Grade by Parental Education")
plt.xlabel("Father's Education (Fedu)")
plt.ylabel("Mother's Education (Medu)")
plt.tight_layout()
plt.show()

# 4. Extracurricular activities
display(df.groupby("activities")["G3"].agg(["count", "mean", "median"]))

plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="activities", y="G3")
plt.title("Final Grade by Extracurricular Activities")
plt.tight_layout()
plt.show()

# 5. Gender comparison
display(df.groupby("sex")["G3"].agg(["count", "mean", "median"]))

plt.figure(figsize=(7, 5))
sns.boxplot(data=df, x="sex", y="G3")
plt.title("Final Grade by Gender")
plt.tight_layout()
plt.show()

# 6. Previous grades vs final grade
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x="G1", y="G3", label="G1")
sns.scatterplot(data=df, x="G2", y="G3", label="G2")
plt.title("Previous Grades vs Final Grade")
plt.xlabel("Previous Grade (G1 or G2)")
plt.ylabel("Final Grade (G3)")
plt.tight_layout()
plt.show()

# 7. Outlier detection using IQR
for col in ["G3", "absences"]:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col}: {len(outliers)} potential outlier rows")
```
